```r
#Limpiamos el entorno de trabajo
rm(list = ls())
```

```r
#Cargamos las librerías necesarias! (Otro día tendré que hacer un archivo para explicar las librerías que más uso)
if (!require("pacman")) install.packages("pacman")
pacman::p_load(tidyverse, srvyr, openxlsx, readr, janitor, dplyr, stringr, googlesheets4, writexl)
```

```r
# ===== CONCENTRADO =========

url_csv <- "https://docs.google.com/spreadsheets/d/1QsgjKt3mS_OV3suheKVu5Dy6NmwQtvyP/export?format=csv&gid=469847885"

concentrado <- read.csv(url_csv, skip = 2) %>%
  clean_names()

head(concentrado)

```

```r
#=========== CONCENTRADO DE ENVÍOS DE ABRIL Y MAYO ======

url2_csv <- "https://docs.google.com/spreadsheets/d/14bp0fler6xTbOQY8YO_ETipTOp7qwiF5/export?format=csv&gid=501566719"

envios <- read.csv(url2_csv) %>%
  clean_names()

head(envios)
```

```r
names(concentrado)
```

```r
#Limpieza de bases

l_concentrado <- concentrado %>%
  rename(estatus_constancia = estatus_constancia_esta_columna_no_se_llena_por_el_personal,
          nombre = nombre_de_la_persona_servidora_publica,
          capacitacion = nombre_de_la_accion_de_capacitacion) %>%
  select(estatus_constancia, nombre, estatus, capacitacion) %>%
  filter(estatus == "Concluido con constancia")

tail(l_concentrado)
```

```r
names(envios)
```

```r
#===== Limpieza de envios ==
l_envios <- envios %>%
  rename(capacitacion = nombre_de_la_accion_de_capacitacion,
        nombre = nombre_de_la_persona_servidora_publica)
head(l_envios)
```

```r
names(envios)
```

```r
# Quitar del concentrado las que no estan en envios
nuevos_registros <- anti_join(l_concentrado, l_envios, by=c("capacitacion", "nombre"))

print(nuevos_registros)
```

```r
#Exportamos la nueva base

write_xlsx(nuevos_registros, "nuevos_registros.xlsx")
```

```r
#========== Personas con capacitación registrada como concluida
#           pero no ha mandado su constancia


faltantes <- concentrado %>%
  rename(estatus_constancia = estatus_constancia_esta_columna_no_se_llena_por_el_personal,
        capacitacion = nombre_de_la_accion_de_capacitacion,
        nombre= nombre_de_la_persona_servidora_publica) %>%
  select(estatus_constancia, estatus, nombre, capacitacion) %>%
  filter(estatus_constancia == "No se ha enviado ")
faltantes
```

```r
# Exportamos faltantes de envíos
write_xlsx(faltantes, "Faltantes.xlsx")
```
