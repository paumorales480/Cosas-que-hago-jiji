```r
#Base de envío anterior (abril)

url_csv <- "https://docs.google.com/spreadsheets/d/1syM1Z-VbpOmg9WvrzoX9482Pjv130-TM/export?format=csv&gid=501566719"

abril <- read.csv(url_csv) %>%
  clean_names() %>%
  select(nombre_de_la_accion_de_capacitacion, nombre_de_la_persona_servidora_publica) %>%
  rename(capacitacion = nombre_de_la_accion_de_capacitacion,
         persona = nombre_de_la_persona_servidora_publica)
head(abril)
```

```r
#Base concentrado

url_csv2 <- "https://docs.google.com/spreadsheets/d/1QsgjKt3mS_OV3suheKVu5Dy6NmwQtvyP/export?format=csv"

concentrado <- read.csv(url_csv2, skip = 2) %>%
  clean_names() %>%
  select(nombre_de_la_accion_de_capacitacion, nombre_de_la_persona_servidora_publica, estatus, estatus_constancia_esta_columna_no_se_llena_por_el_personal) %>%
  rename(capacitacion = nombre_de_la_accion_de_capacitacion,
         persona = nombre_de_la_persona_servidora_publica,
         estatus_constancia = estatus_constancia_esta_columna_no_se_llena_por_el_personal) %>%
  filter(estatus == "Concluido con constancia")
head(concentrado)
```

```r
#Base con la actualización de capacitaciones terminadas pero que no han mandado su constancia
nuevo_mes_faltantes_envio <- anti_join(concentrado, abril, by = c("capacitacion", "persona")) %>%
  filter(estatus_constancia != "Recibida") %>%
  select(persona, capacitacion)
nuevo_mes_faltantes_envio
```

```r
#Nuevo mes (hayan o no mandado su constancia)
nuevo_mes <- anti_join(concentrado, abril, by = c("capacitacion", "persona"))
nuevo_mes
```

```r
write_xlsx(nuevo_mes, "nuevo_mes_colab.xlsx")
#Para descargarla, debo hacer click en la carpeta que está justo a la izquierda
#de esta ventana y ahí debe estar
```

```r
#Nuevo mes con constancias enviadas
nuevo_mes_final <- anti_join(concentrado, abril, by = c("capacitacion", "persona")) %>%
  filter(estatus_constancia=="Recibida")
nuevo_mes_final
```

```r
#BASE FINAL CON CORROBORACIÓN DE ENVÍO DE CONSTANCIAS
write_xlsx(nuevo_mes_final, "nuevo_mes_FINAL_colab.xlsx")
#Para descargarla, debo hacer click en la carpeta que está justo a la izquierda
#de esta ventana y ahí debe estar
```
